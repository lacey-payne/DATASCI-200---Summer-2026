````python
# Q6-2 Grading Tag

import collections

scores = {
    "a": 1, "c": 3, "b": 3, "e": 1, "d": 2, "g": 2,
    "f": 4, "i": 1, "h": 4, "k": 5, "j": 8, "m": 3,
    "l": 1, "o": 1, "n": 1, "q": 10, "p": 3, "s": 1,
    "r": 1, "u": 1, "t": 1, "w": 4, "v": 4, "y": 4,
    "x": 8, "z": 10
}

WORDS = read_sowpods(SOWPODS_PATH)

def score_word(word: str) -> int:
    """Return the Scrabble score for a word using the scores dictionary; wildcards '*' count as 0."""
    return sum(scores.get(ch.lower(), 0) for ch in word if ch != "*")

def validate_rack(rack: str) -> bool:
    """Return True if rack is a valid Scrabble rack (2-7 characters, only A-Z/a-z or *, max two *), False otherwise."""
    if not isinstance(rack, str):
        return False
    if not (2 <= len(rack) <= 7):
        return False
    if any(not (ch.isalpha() or ch == "*") for ch in rack):
        return False
    if rack.count("*") > 2:
        return False
    return True

def run_scrabble(rack: str, words: list[str]) -> dict[str, int]:
    """Return dict mapping uppercase valid words to their highest possible score
    using tiles in rack. Wildcards ('*') represent any letter but score 0.
    Validate rack before calling this function; validation is not performed here.
    This approach is robust against racks with 0, 1, or 2 wildcards."""
    rack_up = rack.upper()
    rack_counts = collections.Counter(rack_up)
    wildcards = rack_counts.get("*", 0)
    results: dict[str, int] = {}

    for word in words:
        w = word.upper()
        # Skip words longer than the rack
        if len(w) > len(rack_up):
            continue

        word_counts = collections.Counter(w)

        # Determine deficits: letters needed beyond what real tiles provide
        deficits = {}
        wild_needed = 0
        for letter, need in word_counts.items():
            have = rack_counts.get(letter, 0)
            if have < need:
                deficit = need - have
                deficits[letter] = deficit
                wild_needed += deficit

        # Skip word if it requires more wildcards than we have
        if wild_needed > wildcards:
            continue

        # Compute the full score assuming all letters are real tiles
        full_score = sum(scores.get(ch.lower(), 0) for ch in w)

        # If wildcards are used, assign them to the lowest-value missing letters
        # to maximize the final score
        if wild_needed > 0:
            needed_letter_values = []
            for letter, cnt in deficits.items():
                needed_letter_values.extend([scores.get(letter.lower(), 0)] * cnt)
            needed_letter_values.sort()
            # Subtract the smallest values (those letters are covered by wildcards, scoring 0)
            subtract = sum(needed_letter_values[:wild_needed])
            final_score = full_score - subtract
        else:
            final_score = full_score

        # Keep the highest score if this word can be formed multiple ways
        if w not in results or final_score > results[w]:
            results[w] = final_score

    return results

if __name__ == "__main__":
    rack = input("Enter your Scrabble rack (2-7 letters, * for wildcard): ").strip()
    if not validate_rack(rack):
        print("Invalid rack. Please use 2-7 characters (letters A-Z or *), and no more than two *.")
    else:
        out = run_scrabble(rack, WORDS)
        if not out:
            print("No valid words can be formed from that rack.")
        else:
            for word, score in sorted(out.items(), key=lambda x: -x[1]):
                print(f"  {word}: {score}")
````
